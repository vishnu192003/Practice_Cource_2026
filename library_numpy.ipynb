```python
## flatten function is used to convert multi-dimensional array into 1 dimensional array
## It returns a copy of the array collapsed into one dimension

v= np.array([[10,20,30],[40,50,60]])
print(v)
print(v.ndim)

v1 = v.flatten()
print(v1)
print(v1.ndim)
```
