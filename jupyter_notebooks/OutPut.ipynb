```python
##  string operator with replacement {} 

name= 'vishnu'
address='nandanvan'
salary=66000.00

print("hello {} your address is {} and your salary {} and we are waiting for our approvel".format(name,address,salary))
```

```python
##  string operator with replacement {} 

name= 'vishnu'
address='nandanvan'
salary=66000.00

print("hello {0} your address is {1} and your salary {2} and we are waiting for our approvel".format(name,address,salary))
```

```python
##  string operator with replacement {} 

name= 'vishnu'
address='nandanvan'
salary=66000.00

print("hello {x} your address is {y} and your salary {z} and we are waiting for our approvel".format(x=name,y=address,z=salary))
```

```python

name= 'vishnu'
address='nandanvan'
salary=66000.00

print(f"hello {name} your address is {address} and your salary {salary} and we are waiting for our approvel")
```
