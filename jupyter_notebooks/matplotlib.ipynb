```python
import matplotlib.pyplot as plt

# Line Chart
x = [1, 2, 3, 4, 5]
y = [10, 25, 30, 44, 50] 

plt.plot(x, y, color='m', label='% of data')
plt.xlabel('X-AXIS')
plt.ylabel('Y-AXIS')
plt.title("Data")
plt.legend()  # Label dikhane ke liye yeh line zaruri hai
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# marker == dot create line
# markersize == dot size can be change

x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 30, 44, 80]

plt.plot(x, y, color='r', label='% of data', marker='o', markersize=8)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
## linewidth to increase width of line

x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 30, 44, 80]

plt.plot(x, y, color='r', label='% of data', marker='o', markersize=8, linewidth= 5)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
## linestyle to increase width of line

x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 30, 44, 80]

plt.plot(x, y, color='r', label='% of data', marker='o', markersize=8, linewidth= 5, linestyle='--', alpha= 0.2)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# random -- module --- rand method
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y)
plt.show()
```

```python
# random -- module --- rand method

x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y,color='m' , alpha=0.4)
plt.show()
```

```python
## histogram 

x = np.random.rand(1000)
plt.hist(x, bins=30, color='y', alpha=0.3, edgecolor='orange')

plt.show()
```

```python
## pie chart 

label = ['java', 'python', 'DA', 'React']
values = [50, 40, 20, 70]
plt.pie(values, labels=label)
plt.show()
```

```python
# autopct ----- auto percentage  
# how to save pie chart davefig

label = ['java', 'python', 'DA', 'React']
values = [50, 40, 20, 70]
plt.pie(values, labels=label, autopct='%1.2f%%')
plt.savefig('Piechart.png')
plt.show()
```
