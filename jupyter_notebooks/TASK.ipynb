## 1

```python
print('1.whitespace remove')
s1 = "   MaChInE LeaRNinG   "
r1 = s1.strip()
print(r1)

print("2.Title Case")
s2 = "MaChInE LeaRNinG"
r2 = s2.title()
print(r2)

print('3.Replace')
s3 = "Machine Learning"
r3 = s3.replace("Learning", "Vision")
print(r3)

print("4.Count")
s4 = "Machine Vision"
r4 = s4.count("i")
print(r4)

print("5.Endswith()")
s5 = "Machine Vision"
r5 = s5.endswith("Vision")
print(r5)
```

## 2

```python
print("1.Int")
a = 10
print(a)
print(type(a))
print(id(a))

print("2.Float")
b = 20.5
print(b)
print(type(b))
print(id(b))

print("3.Complex")
c = 1 + 2j
print(c)
print(type(c))
print(id(c))

print("4.Boolean")
d = True
print(d)
print(type(d))
print(id(d))


print("5. List")
e = [1, 2, 3]
print(e)
print(type(e))
print(id(e))

print("6.Tuple")
f = (4, 5, 6)
print(f)
print(type(f))
print(id(f))

```

## 3

```python
print('1.isalpha()')
a = "Hello"
print(a.isalpha())

a = "Hello78"
print(a.isalpha())

print('2.isdigit()')
b = "123"
print(b.isdigit())

b = "123av1"
print(b.isdigit())

print('3.isalnum()')
c = "Hello123"
print(c.isalnum())

c = "Hello123 78@"
print(c.isalnum())
```

## 4

```python
course = "ArtificialIntelligence"

print("Print first letter")
print(course[0])         

print("Print last letter ")
print(course[-1])         

print("Extract Artificial")
print(course[0:10])       

print("Extract Intelligence")
print(course[10:])       

print("Reverse the string")
print(course[::-1])       
```
