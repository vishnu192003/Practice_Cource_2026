```python
# 53.	*****
#       *   *
#       *   *
#       *   *
#       *****

n = int(input("Enter Number: "))

for i in range(1, n + 1):
    if i == 1 or i == n: 
        print("* " * n)
    else:
        print("* " + "  " * (n - 2) + "*")  
        
```

```python
# 54.	Hollow Triangle
# *
# **
# * *
# *  *
# *****

n = int(input("Enter Number: "))

for i in range(1, n + 1):
    if i==1  :
        print("* " * i)
    if i==n:
        print("* " * i)
    else:
        print("* " + "  " * (i - 2) + "*")  
```
