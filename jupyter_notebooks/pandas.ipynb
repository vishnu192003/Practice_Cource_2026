```python
import pandas as pd
import numpy as np

# Student dataset with 5 records (including missing values)
data = {
    'seating_': [101, 102, 103, 104, 105],
    'Name': ['Vishnu', 'Sarthak', 'Sanket', 'Aarav', 'Priya'],
    'Age': [85, 92, np.nan, 88, 95],
    'Branch': ['CS', 'IT', 'ECE', 'ME', np.nan]
}

# Create the DataFrame
df3 = pd.DataFrame(data)
df3
```

```python
df3['Name'] = df3['Name'].replace('Priya','Aarav')
```

```python
df3
```

```python
df3['Age'] = df3['Age'].replace(95.0,88.0)
```

```python
df3
```

```python
df3['Branch'] = df3['Branch'].replace('Nan','ME')
```

```python
df3
```

```python
df3['seating_'] = df3['seating_'].replace(105,104)
```
