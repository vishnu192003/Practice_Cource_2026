## Argument 

```python
# Explanation for default argument: 
# -- Default arguments are values that are provided while defining functions.
# -- The assignment operator = is used to assign a default value to the argument.
# -- Default arguments become optional during the function calls.
# -- If we provide a value to the default arguments during function calls, it overrides the default value.
# -- The function can have any number of default arguments.
# -- Default arguments should follow non-default arguments.


#1.positional
#2.keyword
#3.default
#4.Arbitary
```

```python
### reverse a string using inxing,if ,return 

def rev(name):
    if len(name)==0:
        return name
    return name[-1]+rev(name[:-1])
rev("vishnu")

```

## Functions Scope(local & global variable)

```python
# LOCAL SCOPE ----> variable inside a function

def sco():
    a=10
    print("value of a : ",a)
sco()
```

```python
# GLOBAL SCOPE -----> variable outside any function

def sco():
    a=10
    print("value of a : ",a)
sco()
a=20
print("value of a : ",a)
```

```python
a=20
def sco():
    a=10
    print("value of a : ",a)
    print("value of a : ",a)
sco()

print("value of a : ",a)
```

```python
## nested function 

b=20
def outer():
    a=10
    print("this is outer function")
    print(a)
    def inner():
        print("this is a inner function" )
        print(a)
        print(a)
        print(b)
    inner()
outer()
print("value of b" ,b)  
```

```python
# global variable


def outer():
    global a 
    a=10
    print(a)
   
outer()
print("value of",a)
```

```python
balance=10000


while True:

    print("""
    1.Check Balance
    2.Deposit
    3.Withdraw
    4.Exit
    """)


    choice=int(input("Choose option: "))


    if choice==1:

        print(balance)


    elif choice==2:

        amount=int(input("Deposit amount: "))

        balance+=amount


    elif choice==3:

        amount=int(input("Withdraw amount: "))


        if amount<=balance:

            balance-=amount

        else:

            print("Insufficient Balance")


    elif choice==4:

        break


    else:

        print("Invalid Choice")
```
