```python
# User se number input lo
n = int(input("Enter Number: "))

# ---------- Upper Half Pyramid ----------
for i in range(1,n):        # Rows 1 se n-1 tak chalengi

    for j in range(0, n-i):  # Har row ke shuru me spaces print karo
        print(end=' ')

    for j in range(0, i):    # Har row me i stars print karo
        print('*', end=' ')

    print()                  # Next line me jao


# ---------- Lower Half Pyramid ----------
for i in range(n, 0, -1):    # Rows n se 1 tak reverse me chalengi

    for j in range(0, n-i):  # Har row ke shuru me spaces print karo
        print(end=' ')

    for j in range(0, i):    # Har row me i stars print karo
        print('*', end=' ')

    print()                  # Next line me jao
```

```python
# User se number input lo
n = int(input("Enter Number: "))

# ================== Upper Half Pyramid ==================

# i row ko represent karta hai.
# range(1, n) ka matlab 1 se lekar n-1 tak chalega.
for i in range(1, n):

    # Ye loop spaces print karta hai.
    # Jitni row badhegi, utni spaces kam hongi.
    # Example: n=6
    # i=1 → 5 spaces
    # i=2 → 4 spaces
    # i=3 → 3 spaces
    for j in range(0, n - i):
        print(end=' ')

    # Ye loop stars print karta hai.
    # Har row me i ke barabar stars print honge.
    # Example:
    # i=1 → *
    # i=2 → * *
    # i=3 → * * *
    for j in range(0, i):
        print('*', end=' ')

    # Ek row complete hone ke baad next line me jao.
    print()


# ================== Lower Half Pyramid ==================

# Ye reverse loop hai.
# range(n, 0, -1) ka matlab n se 1 tak aana.
# Example: n=6
# i = 6, 5, 4, 3, 2, 1
for i in range(n, 0, -1):

    # Har next row me spaces badhti jayengi.
    # Example:
    # i=6 → 0 spaces
    # i=5 → 1 space
    # i=4 → 2 spaces
    for j in range(0, n - i):
        print(end=' ')

    # Har row me i ke barabar stars print honge.
    # Example:
    # i=6 → * * * * * *
    # i=5 → * * * * *
    # i=4 → * * * *
    for j in range(0, i):
        print('*', end=' ')

    # Next line me jao.
    print()
```
