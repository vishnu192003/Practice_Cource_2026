## 1. Employee Salary Calculator

Scenario based interview problem.

Difficulty: ⭐⭐

Test case: Try multiple inputs.

```python
salary=float(input('Salary: '))
final=salary+salary*.2+salary*.1-salary*.05
print(final)
```

## 2. Electricity Bill

Scenario based interview problem.

Difficulty: ⭐⭐

Test case: Try multiple inputs.

```python
u=int(input('Units: '))
if u<=100: b=u*5
elif u<=300: b=500+(u-100)*7
else: b=1900+(u-300)*10
print(b)
```

## 3. Grade Calculator

Scenario based interview problem.

Difficulty: ⭐⭐

Test case: Try multiple inputs.

```python
m=int(input('Marks: '))
if m>=90: print('A')
elif m>=75: print('B')
elif m>=50: print('C')
else: print('Fail')
```

## 4. Prime Checker

Scenario based interview problem.

Difficulty: ⭐⭐

Test case: Try multiple inputs.

```python
n=int(input('Number: '))
print('Prime' if n>1 and all(n%i for i in range(2,n)) else 'Not Prime')
```

## 5. Palindrome Checker

Scenario based interview problem.

Difficulty: ⭐⭐

Test case: Try multiple inputs.

```python
s=input('Text: ')
print('Palindrome' if s==s[::-1] else 'Not Palindrome')
```
