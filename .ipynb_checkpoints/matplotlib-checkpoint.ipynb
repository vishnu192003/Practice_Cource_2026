```python
x = ['A','B','C','D','E']
y = [50,40,30,60,10]

plt.plot(x,y, color='y')
plt.show()
```

```python
## grid()

x = ['A','B','C','D','E']
y = [50,40,30,60,10]

plt.plot(x,y, color='c')
plt.grid()
plt.show()
```

```python
## labled()

x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 30, 44, 50] 


plt.plot(x, y, color='c', label='% of data')
plt.grid()
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ['A', 'B', 'C', 'D', 'E']


y1 = [10, 25, 30, 44, 50] 
y2 = [20, 15, 35, 25, 40] 


plt.plot(x, y1, color='c', label='Data 1')
plt.plot(x, y2, color='m', label='Data 2')


plt.ylabel('y-axis')
plt.xlabel('x-axis')

plt.grid()
plt.legend()  
plt.show()
```

```python

x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 30, 44, 50] 


plt.plot(x, y, color='c', label='% of data')
plt.grid()
plt.ylabel('y-axis')
plt.xlabel('x-axis')
plt.title(''' 
Practice

''')
plt.legend()
plt.show()
```

```python
x = ['A', 'B', 'C', 'D', 'E']
y = [10, 25, 30, 44, 50] 


plt.plot(x, y, color='c', label='% of data')
plt.grid()
plt.ylabel('percentage of data ')
plt.xlabel('x-axis')
plt.title(''' 
Practice

''')
plt.legend()
plt.show()
```
