```python
n = 7
for i in range(n):
    for j in range(n):
        # Checks for the left wall, right wall, or main diagonal
        if j == 0 or j == n - 1 or j == i:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()  # Out of the 'j' loop, inside the 'i' loop
```

```python
n = 5
for i in range(n):
    for j in range(2 * n - 1):
        # Conditions to form the letter 'V'
        if i == j or i + j == 2 * n - 2:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()  # Correctly indented outside the 'j' loop
```

```python
for i in range(1,6):
    print("※" * i)
```

```python
n=5
for i in range (1,n+1):
    print(" "*(n-i)+"⁕ "*i)
for i in range(n-1,0,-1):
    print(" " * (n-i)+"⁕ "*i)
```
