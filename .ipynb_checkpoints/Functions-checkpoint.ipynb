```python
## Addition 
```

```python
def add(a, b): 
  return a + b  # This line sends back sum of a and b 
result = add(3,5)
print(result)      
# Output: 8 
```

```python
def sub(a, b): 
  return a - b  # This line sends back sum of a and b 
result = sub (34,5)
print(result)      
# Output: 8 
```

```python
def mul(a, b): 
  return a * b  # This line sends back sum of a and b 
result = mul (7,5)
print(result)      
# Output: 8 
```

```python
def divide(a, b):   # a,b are 2 parameters 
 return a / b 
result = divide(b=10, a=20) # with keyword arguments 
print(result)    
# Output: 2 
```

## Argument 

```python
# Explanation for default argument: 
# -- Default arguments are values that are provided while defining functions.
# -- The assignment operator = is used to assign a default value to the argument.
# -- Default arguments become optional during the function calls.
# -- If we provide a value to the default arguments during function calls, it overrides the default value.
# -- The function can have any number of default arguments.
# -- Default arguments should follow non-default arguments.


#1.positional
#2.keyword
#3.default
#4.Arbitary
```

```python
# ## Positional Argument
```

```python
# argument passed in order called as positional arguement

def val(name,surname):
    print(f'my name is {name}) and my surname is {surname}')
val("vishnu","chacherkar")
```

```python
# Keyword Argument

def val(name,surname):
    print(f'my name is {name} and my surname is {surname}')
val( surname ="chacherkar", name= "vishnu" )
```

```python
# Default Argument

def pli(a,b=10):
    print(a+b)
pli(10)
```

```python
def pli(a,b=2):
    print(a+b)
pli(10)
```

```python
def pli(a,b,c=10):
    print(a+b)
pli(10,20)
```

```python
#Arbitary Arguments(*args and **kwargs)
```

```python
def add(*v):
    print((v))
add(10,20,30,40)
```

```python
def ab(*argv):
    print(sum(argv))          
ab(10,20,30,40)
```

```python
# lambda function 
# anonymous function defined usoing lambda 

l= lambda n: n*n
print(l(2))
```

```python
p= lambda n,m: n*m
print(p(2,3))
```

```python

# Take 2 parameters, add them, then divide by 2
p = lambda n, m: (n + m) / 2

print(p(2, 3))
```

```python
def msg():
    print("hello")
    return "world"

r=msg()
print(r)
```

```python
## Factorial
def fac(f):
    if f<=1:
        return 1
    return f*fac(f-1)
fac(5)
```

```python
### reverse a string using inxing,if ,return 

def rev(name):
    if len(name)==0:
        return name
    return name[-1]+rev(name[:-1])
rev("vishnu")

```
