```python
# Exercise 18: Find the common items between two arrays
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([3, 4, 5, 6, 7])

common = np.intersect1d(a, b)
print(common)
```

```python
# Exercise 23: Normalize a NumPy array (values between 0 and 1)
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print(normalized)
```

```python
# Exercise 27: Sorting a NumPy array based on a specific column
import numpy as np

arr = np.array([[1, 4, 9], [3, 1, 6], [5, 2, 8]])

# Sort based on 2nd column (index 1)
sorted_arr = arr[arr[:, 1].argsort()]
print(arr)
print()
print( sorted_arr)
```
