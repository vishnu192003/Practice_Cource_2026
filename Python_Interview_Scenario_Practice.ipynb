```python
# 2. Electricity Bill
units=int(input('Units: '))
if units<=100: bill=units*5
elif units<=300: bill=100*5+(units-100)*7
else: bill=100*5+200*7+(units-300)*10
print(bill)
```

```python
# 3. Grade Calculator
marks=int(input('Marks: '))
if marks>=90: print('A')
elif marks>=75: print('B')
elif marks>=50: print('C')
else: print('Fail')
```

```python
# 7. Prime Checker
n=int(input('Number: '))
prime=True
for i in range(2,n):
    if n%i==0:
        prime=False
        break
print('Prime' if prime else 'Not Prime')
```

```python
# 8. Palindrome
s=input('Text: ')
print('Palindrome' if s==s[::-1] else 'Not Palindrome')
```
