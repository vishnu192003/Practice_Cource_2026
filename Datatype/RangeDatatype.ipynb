```python
## Range 
## Range Data type represent sequence of numbers
## the element presennt in  range data type are not modifiable i.e range dataype is immutable
```

```python
re= range(10)
print (type(re))
print(re)
```

```python
for i in re:
    print(i)
```

```python
re= range(1,50)
print (type(re))
print(re)

for i in re:
    print(i)
```

```python
## start: end-1 : increment/ decrement
```

```python
v= range(1,20)
print(type(v))
print(v)
h= v[1:20:2]
for l in h:
   print(l)
```

```python
v= range(1,20)
print(type(v))
print(v)
h= v[20:1:-2]
for l in h:
   print(l)
```
