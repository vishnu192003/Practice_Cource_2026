## byte array 

```python
## bytearray is mutable
## bytearray  support indexing and slicing
## mutable
## (0,256)
```

```python
li= [10,20,30,40,50]
print(li)

a= bytearray(li)
print(type(a))
for i in a:
    print(i)
```

```python
## indexing -----> mutable
```

```python
a[1]= 200
for i in a:
    print(i)
```
