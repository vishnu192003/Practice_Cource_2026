## TUPLE () 


```python
## tuple datatype is represented in parenthesis () with , separator
## tuple is immutable
## Allows heterogeneous. 
## Indexing and slicing.
## Support. Insertion object.

```

```python
t= ()
print(t)
print(type(t))
```

```python
# float value
tu = ( 10.8952659 ,)
print(tu)
print(type(tu))
```

```python
## duplicate

tup = ( 10.8952659 ,1,1,1,1,1,1,2,1)
print(tup)
print(type(tup))
```

```python
## hretrogeneous

tupl = ( 10.8952659 , 45+125j, 'vishnu' , True )
print(tupl)
print(type(tupl))
```

```python
## immutable

ctu = ( 10.8952659 , 'it' , 'vishnu ', 'standard ', True, 6.66)
ctu[2]= "hello"
print(ctu)
print(type(ctu))
```

```python
t =(1,2,3,4,5,6)
print(id(t))
t=t*2
print(t)
print(id(t))
```

```python
t=()
d = t.append(10)
print(d)
```

```python
## indexing

ctu1= (10,20,30,40,50,60,70,80,90)
print(ctu1[4])
```

```python
print(ctu1[3:7])
print(ctu1[0:9:2])
```

```python
## how to convert other datatype
l= [10,20,30,40,50,60,70,80,90]
print(type(l))
t=tuple(l)
print(t)
print(type(t))
```

```python
## sorted 
s= ( 60, 70, 10, 20, 30, 40, 50, 60, 70, 10, 20, 30, 40)
print(sorted(s))
```
