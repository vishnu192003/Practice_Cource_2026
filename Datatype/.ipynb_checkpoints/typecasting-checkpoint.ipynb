## typecasting

```python
# 1) int() 
# 2) float() 
# 3) complex() 
# 4) bool() 
# 5) str() 
```

## int()

```python
## we can convert from any type to int except complex type
```

```python
## float () ---> int ()
print("float datatype into int datatype: " , int(1264.6565))
```

```python
## float () ---> int ()
print("float datatype into int datatype: " , int(10.00))
```

```python
## complex() ---->int

print("complex datatype into int datatype: " , int(10+5j))
```

```python
## bool() ---> int

print("bool datatype into int datatype: " , int(True))
```

```python
## bool() ---> int

print("bool datatype into int datatype: " , int(False))
```

```python
## str ----int()

# if we want to convert str type to int type
# compalsoury string should contain only integral value and
# should be specified in base -10.

```

```python
print("string datatype into int datatype: " , int('10'))
```

```python
print("string datatype into int datatype: " , int('10'))
```

## float()


```python
## float()
# we can convert any type of value to float except complex type
```

```python
print("int datatype into float datatype: " , float(10))
```

```python
print("int datatype into float datatype: " , float(0b01011101))
```

```python
print("int datatype into float datatype: " , float(0o01011101))
```

```python
print("int datatype into float datatype: " , float(0xe011101))
```

```python
## complex to float ()
```

```python
print("complex datatype into float datatype: " , float(10+4j))
```

```python
print("bool datatype into float datatype: " , float(True))
```

```python
print("bool datatype into float datatype: " , float(False))
```

```python
## string to float 

# whenever we are trying to convert str type to convert float type
# compulsoury str should be either integral or floating literal and
# should be specify only based -10
```

```python
print("string datatype into int datatype: " , float ('10'))
```

```python
print("string datatype into int datatype: " , float ('0b010110'))
```
