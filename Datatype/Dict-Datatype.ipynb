```python
## dict datatype is represent {} with the key and value pair 
## dict does not allow duplicate key but value can be duplicate
## dict is mutable
## dict order insertion not allowed
## dictionary does not support indexing and sclicing
```

```python
## syntax of dictionary
## {key:values}
## {'name': 'vishnu', 'add': 'nagpur' , 'house_no': 123}
```

```python
di={'name': 'vishnu', 'add': 'nagpur' , 'house_no': 123}
print(di)
print(type(di))
```
