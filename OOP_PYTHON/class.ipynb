```python
# local variable

class Msg:
    def display(self):
        str="hello world"
        s = 1000
        print(str)
        print(s)

ob=Msg()
ob.display()
```

```python
# Instance Variable

class Instance:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
        print(name)
        print(salary)
    def show(self):
        print(self.name)
        print(self.salary)

ob=Instance("vishnu",541213.00)
ob.show()
```

```python
# static variable

class Stack:

    cname="sofrtonix"
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
        print(name)
        print(salary)
    def show(self):
        print(self.name)
        print(self.salary)
        print(self.cname)

ob=Stack("vishnu",541213.00)
ob.show()
```

```python
# Task College
class CollegeStudent:
    college_name = "V Group of Institutions"

    def __init__(self, name, roll_no):
        
        self.name = name
        self.roll_no = roll_no

    def show(self):
        print("College:", CollegeStudent.college_name)
        print("Student Name:", self.name)
        print("Roll No:", self.roll_no)
        print("-"*30)


student = CollegeStudent("Sarthak", 101)
student.show()
student = CollegeStudent("Harsh", 102)
student.show()
student = CollegeStudent("srushti", 103)
student.show()
```

```python
# bank 

# Static Variable (Class Variable)

class BankAccount:
  
    bank_name = "State Bank of India"

    def __init__(self, name, balance):
      
        self.name = name
        self.balance = balance        

    def show(self):
        print("Bank:", BankAccount.bank_name)
        print("Name:", self.name)
        print("Balance: ₹", self.balance)
        print("-"*30)

account1 = BankAccount("sarthak", 50000)
account2 = BankAccount("harsh", 85000)
account3 = BankAccount("srusti", 70000)

account1.show()
account2.show()
account3.show()

```
