```python
## identity operator 
## is ---> address check 

t1=(10,20,30,40,50)
t2=(10,20,30,40,50)
print(id(t1))
print(id(t2))
print(t1 is t2)
print(t1 == t2)
```

## HM assignment operator

```python
# = Assignment Operator

a = 10
print(a)
print('------------------------')


# += Add and Assign

b = 10
b += 5
print(b)
print('------------------------')

# -= Subtract and Assign

c = 10
c -= 3
print(c)
print('------------------------')

# *= Multiply and Assign

d = 10
d *= 2
print(d)
print('------------------------')

# /= Divide and Assign

e = 10
e /= 2
print(e)
print('------------------------')

# %= Modulus and Assign

f = 10
f %= 3
print(f)
print('------------------------')

# //= Floor Divide and Assign

g = 10
g //= 3
print(g)
print('------------------------')

# **= Power and Assign

h = 2
h **= 3
print(h)
print('------------------------')

# &= Bitwise AND and Assign

i = 5
i &= 3
print(i)
print('------------------------')

# |= Bitwise OR and Assign

j = 5
j |= 3
print(j)
print('------------------------')

# ^= Bitwise XOR and Assign

k = 5
k ^= 3
print(k)
print('------------------------')

# >>= Right Shift and Assign

l = 8
l >>= 1
print(l)
print('------------------------')
# <<= Left Shift and Assign

m = 8
m <<= 1
print(m)

```

## HM operator precedence

```python
# 1) () Parenthesis

a = (3 + 2) * 4
print(a)
print('---------------')


# 2) ** Exponential Operator

b = 2 ** 3
print(b)
print('---------------')

# 3) ~, - Bitwise Complement Operator, Unary Minus Operator

c = ~5
d = -5
print(c)
print(d)
print('---------------')

# 4) *, /, %, // Multiplication, Division, Modulo, Floor Division

e = 10 * 2
f = 10 / 2
g = 10 % 3
h = 10 // 3

print(e)
print(f)
print(g)
print(h)
print('---------------')

# 5) +, - Addition, Subtraction

i = 10 + 5
j = 10 - 5

print(i)
print(j)
print('---------------')

# 6) <<, >> Left and Right Shift

k = 5 << 1
l = 5 >> 1

print(k)
print(l)
print('---------------')

# 7) & Bitwise And

m = 5 & 3
print(m)
print('---------------')

# 8) ^ Bitwise X-OR

n = 5 ^ 3
print(n)
print('---------------')
# 9) | Bitwise OR

o = 5 | 3
print(o)
print('---------------')

# 10) >, >=, <, <=, ==, != Relational OR Comparison Operators

p = 10 > 5
q = 10 >= 10
r = 5 < 10
s = 5 <= 5
t = 10 == 10
u = 10 != 5

print(p)
print(q)
print(r)
print(s)
print(t)
print(u)
print('---------------')
# 11) =, +=, -=, *= Assignment Operators

v = 10
v += 5
print(v)

w = 10
w -= 3
print(w)

x = 10
x *= 2
print(x)

print('---------------')
# 12) is, is not Identity Operators

y = [1, 2, 3]
z = y

print(y is z)
print(y is not z)

print('---------------')
# 13) in, not in Membership Operators

a1 = [10, 20, 30]

print(20 in a1)
print(40 not in a1)

print('---------------')
# 14) not Logical NOT

a2 = not True
print(a2)

print('---------------')
# 15) and Logical AND

a3 = True and False
print(a3)

print('---------------')
# 16) or Logical OR

a4 = True or False
print(a4)
```
